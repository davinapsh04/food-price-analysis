```python
c =pangan2.corr()
c
```

```python
#Membuat visualisasi korelasi antar variabel
fig = plt.subplots(10, 10, figsize=(35,30))
plt.subplots_adjust(left=0.1, bottom=0.1, right=0.9, top=0.9, wspace=0.4, hspace=0.4)
col = c.columns
k = 1
x = 0
for i in col:
  y = 0
  for j in col:
    plt.subplot(10, 10, k)
    plt.scatter(pangan[i],pangan[j], color = colors[k])
    rho = c.iloc[x][y]
    plt.title('r = ' + f'{rho:.2f}')
    plt.xlabel(i)
    plt.ylabel(j)
    k = k+1
    y = y+1
  x = x+1
plt.suptitle("Visualisasi Korelasi Antar Variabel")
plt.show()
```

```python
#Cara 1: Menampilkan visualisasi korelasi antar pangan
hm = sns.heatmap(data=c, annot=True)
plt.title("Korelasi Antar Variabel")
plt.xticks(range(len(c.columns)), c.columns, rotation=90)
plt.yticks(range(len(c.columns)), c.columns)
plt.show()
```

```python
#Cara 2: Menampilkan visualisasi korelasi antar pangan
#Pilihan warna dapat diakses melalui https://matplotlib.org/stable/users/explain/colors/colormaps.html
fig, ax = plt.subplots()
im = ax.imshow(c, cmap = 'summer')

plt.xticks(range(len(c.columns)), c.columns, rotation=90)
plt.yticks(range(len(c.columns)), c.columns)

plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")

for i in range(len(c)):
    for j in range(len(c)):
        text = ax.text(j, i, '%.2f' % c.iloc[i][j], ha="center", va="center", color="b")

ax.set_title("Korelasi Antar Pangan")
fig.tight_layout()
plt.show()
```

```python
komoditas = ['Beras','Daging Ayam','Telur Ayam','Bawang Merah',
             'Bawang Putih','Cabai Merah','Cabai Rawit Hijau',
             'Cabai Rawit Merah','Minyak Goreng','Gula Pasir']
```

```python
annual_mean = pangan.groupby('Tahun')[komoditas].mean()
```

```python
annual_nominal_change = annual_mean.diff().mean().round(2)
annual_percent_change = (annual_mean.pct_change() * 100).mean().round(2)
```

```python
print("Rata‑rata kenaikan / penurunan per tahun (nominal, IDR):")
print(annual_nominal_change)
```

```python
print("\nRata‑rata % perubahan per tahun (Year of Year):")
print(annual_percent_change)
```

```python
monthly_mean = pangan.groupby('Bulan')[komoditas].mean()
```

```python
monthly_nominal_change = monthly_mean.diff().mean().round(2)
monthly_percent_change = (monthly_mean.pct_change() * 100).mean().round(2)
```

```python
print("Rata‑rata kenaikan / penurunan per bulan (nominal, IDR):")
print(monthly_nominal_change)
```

```python
print("\nRata‑rata % perubahan per bulan (Month Of Month):")
print(monthly_percent_change)
```
